```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    while queue:
        node = queue.popleft()

        if node not in visited:
            print(node, end=" ")
            visited.add(node)

            # Add unvisited neighbours to the queue
            for neighbour in graph[node]:
                if neighbour not in visited:
                    queue.append(neighbour)


# Graph represented using a dictionary
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

print("BFS Traversal:")
bfs(graph, 'A')
```

```python
def dfs(graph, node, visited):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)

        for neighbour in graph[node]:
            dfs(graph, neighbour, visited)


# Graph represented using a dictionary
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

visited = set()

print("DFS Traversal:")
dfs(graph, 'A', visited)
```

```python
# Greedy Best-First Search

import heapq

def greedy_best_first_search(graph, heuristic, start, goal):
    visited = set()
    priority_queue = []

    # (heuristic value, node)
    heapq.heappush(priority_queue, (heuristic[start], start))

    while priority_queue:
        h, current = heapq.heappop(priority_queue)

        if current in visited:
            continue

        print(current, end=" ")
        visited.add(current)

        # Goal reached
        if current == goal:
            print("\nGoal reached!")
            return

        # Add neighbours based on heuristic value
        for neighbour in graph[current]:
            if neighbour not in visited:
                heapq.heappush(
                    priority_queue,
                    (heuristic[neighbour], neighbour)
                )

    print("\nGoal not found!")


# Graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

# Heuristic values
heuristic = {
    'A': 5,
    'B': 3,
    'C': 2,
    'D': 4,
    'E': 1,
    'F': 0
}

print("Greedy Best-First Search:")
greedy_best_first_search(graph, heuristic, 'A', 'F')
```
