```python
from queue import PriorityQueue

def a_star(graph, heuristic, start, goal):
    open_set = PriorityQueue()
    open_set.put((0, start))

    g_cost = {start: 0}
    parent = {start: None}

    while not open_set.empty():
        current = open_set.get()[1]

        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]
            return path[::-1], g_cost[goal]

        for neighbor, cost in graph[current]:
            new_g = g_cost[current] + cost

            if neighbor not in g_cost or new_g < g_cost[neighbor]:
                g_cost[neighbor] = new_g
                f_cost = new_g + heuristic[neighbor]
                open_set.put((f_cost, neighbor))
                parent[neighbor] = current

    return None, float("inf")


# Graph represented as adjacency list
graph = {
    'A': [('B', 1), ('C', 3)],
    'B': [('D', 3), ('E', 1)],
    'C': [('F', 5)],
    'D': [('G', 2)],
    'E': [('G', 4)],
    'F': [('G', 1)],
    'G': []
}

# Heuristic values (estimated cost to goal)
heuristic = {
    'A': 6,
    'B': 4,
    'C': 4,
    'D': 2,
    'E': 2,
    'F': 1,
    'G': 0
}

start = 'A'
goal = 'G'

path, cost = a_star(graph, heuristic, start, goal)

if path:
    print("Shortest Path:", " -> ".join(path))
    print("Total Cost:", cost)
else:
    print("No path found.")
```

```python
from queue import PriorityQueue

def uniform_cost_search(graph, start, goal):
    pq = PriorityQueue()
    pq.put((0, start))

    visited = set()
    parent = {start: None}
    cost = {start: 0}

    while not pq.empty():
        current_cost, current_node = pq.get()

        if current_node in visited:
            continue

        visited.add(current_node)

        if current_node == goal:
            path = []
            while current_node is not None:
                path.append(current_node)
                current_node = parent[current_node]
            return path[::-1], current_cost

        for neighbor, edge_cost in graph[current_node]:
            new_cost = current_cost + edge_cost

            if neighbor not in cost or new_cost < cost[neighbor]:
                cost[neighbor] = new_cost
                parent[neighbor] = current_node
                pq.put((new_cost, neighbor))

    return None, float("inf")


# Graph represented as an adjacency list
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 3)],
    'D': [('G', 1)],
    'E': [('G', 2)],
    'F': [('G', 2)],
    'G': []
}

start = 'A'
goal = 'G'

path, total_cost = uniform_cost_search(graph, start, goal)

if path:
    print("Shortest Path:", " -> ".join(path))
    print("Total Cost:", total_cost)
else:
    print("No path found.")
```

```python
from queue import PriorityQueue

def dijkstra(graph, start, goal):
    pq = PriorityQueue()
    pq.put((0, start))

    distance = {node: float('inf') for node in graph}
    distance[start] = 0

    parent = {start: None}

    while not pq.empty():
        current_distance, current_node = pq.get()

        if current_node == goal:
            break

        for neighbor, weight in graph[current_node]:
            new_distance = current_distance + weight

            if new_distance < distance[neighbor]:
                distance[neighbor] = new_distance
                parent[neighbor] = current_node
                pq.put((new_distance, neighbor))

    # Reconstruct path
    if distance[goal] == float('inf'):
        return None, float('inf')

    path = []
    current = goal
    while current is not None:
        path.append(current)
        current = parent[current]

    return path[::-1], distance[goal]


# Graph represented as adjacency list
graph = {
    'A': [('B', 2), ('C', 4)],
    'B': [('C', 1), ('D', 7)],
    'C': [('E', 3)],
    'D': [('F', 1)],
    'E': [('D', 2), ('F', 5)],
    'F': []
}

start = 'A'
goal = 'F'

path, cost = dijkstra(graph, start, goal)

if path:
    print("Shortest Path:", " -> ".join(path))
    print("Total Cost:", cost)
else:
    print("No path found.")
```
